```python
#Etape 1
import json

with open ("../data/villes.json", encoding="utf-8") as f:
    data_villes = json.load(f)

print(type(data_villes))
print(f"Nombre de ville : {len(data_villes["villes"])}")
print(f"Nom de la deuxième ville : {data_villes["villes"][1]["nom"]}")
print(f"Nombre de restaurants à Syracuse : {len(data_villes["villes"][2]["restaurants"])}")
print(f"Nom du premier restaurant de Palerme : {data_villes["villes"][1]["restaurants"][0]["nom"]}")
```

```python
#Etape 2
for ville in data_villes["villes"]:
    print(f"Ville : {ville["nom"]} -- nombre de restaurants {len(ville["restaurants"])}")
```

```python
#Etape 3
import pandas as pd
df_villes = pd.json_normalize(data_villes, record_path=["villes","restaurants"], meta=[["villes", "nom"], ["villes","region"],["villes","habitants"]])
df_villes = df_villes.rename(columns={"villes.nom" : "villes", "villes.region" : "region", "villes.habitants" : "habitant"})
print(df_villes)
```

```python
df_villes["note_moyenne"] = df_villes["notes"].apply(lambda x : round(sum(x) / len(x),2) if x else None)
df_villes.head()

df_grouped = round(df_villes.groupby("villes")["note_moyenne"].mean(),2)
df_grouped_sorted = df_grouped.sort_values(ascending = False)
print(f"Ville avec meilleure note moyenne : {df_grouped_sorted.index[0]}")
print("Nombre de type de restaurants tous lieux confondus : ")
print(df_villes["type"].value_counts())
```
