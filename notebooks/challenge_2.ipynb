```python
adresses = [
    "20 avenue de Ségur 75007 Paris",
    "Grand Place, Lille",
    "1 rue de la Paix, Paris",
    "place Bellecour, Lyon",
    "zzzzz qwerty 99999",
]
```

```python
#Etape 1
import requests
import time
import json
headers = {"User-Agent" : "bootcamp-simplon-data/1.0"}
url = "https://api-adresse.data.gouv.fr/search/"
r = requests.get(url, params = {"q": "20 avenue de Ségur 75007 Paris"}, headers = headers)
time.sleep(1)
print(r.status_code)
reponse = r.json()
print(f"Le label du premier resultat est : {r.json()["features"][0]["properties"]["label"]}")
print(f"Le score du premier resultat est : {round(r.json()["features"][0]["properties"]["score"],2)}")

```

```python
print(json.dumps(reponse, ensure_ascii=False, indent=2))
```

```python

headers = {"User-Agent" : "bootcamp-simplon-data/1.0"}
url = "https://api-adresse.data.gouv.fr/search/"
r = requests.get(url, params = {"q": "zzzzz qwerty 99999"}, headers = headers)
time.sleep(1)
print(r.status_code)
reponse = r.json()
print(json.dumps(reponse, ensure_ascii=False, indent=2))
```

```python
#Etape 2
liste_adresses = []
empty = []
for adresse in adresses : 
    headers = {"User-Agent" : "bootcamp-simplon-data/1.0"}
    url = "https://api-adresse.data.gouv.fr/search/"
    r = requests.get(url, params = {"q": adresse}, headers = headers)
    time.sleep(1)
    r.raise_for_status()
    reponse = r.json()
    if reponse["features"] != empty : 
        liste_adresses.append(
            { 
                "adresse" : r.url, 
                "label" : reponse["features"][0]["properties"]["label"],
                "score" : round(r.json()["features"][0]["properties"]["score"],2),
                "postcode" : reponse["features"][0]["properties"]["postcode"],
                "city" : r.json()["features"][0]["properties"]["city"]

            }
        )
    else : 
        liste_adresses.append(
                { 
                    "adresse" : None, 
                    "label" : None,
                    "score" : None,
                    "postcode" : None,
                    "city" : None
    
                }
            )

  
for adresse in liste_adresses : 
    print(json.dumps(adresse, ensure_ascii=False, indent=2))
```
